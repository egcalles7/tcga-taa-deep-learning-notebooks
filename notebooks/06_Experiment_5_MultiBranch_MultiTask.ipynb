# 06 — Experiment 5: Multi-Branch Multi-Task Neural Network

This is the most biologically motivated model. It learns separate representations for clinical/cancer-type, mutation, and copy-number features before fusing them into a shared representation with gene-specific output heads.

```python
from pathlib import Path
import os

def find_project_root(start=None):
    start = Path.cwd() if start is None else Path(start)
    start = start.resolve()
    for candidate in [start] + list(start.parents):
        if (candidate / "config" / "genes.json").exists():
            return candidate
    raise FileNotFoundError("Could not find project root. Start Jupyter from repo root or notebooks folder.")

PROJECT_ROOT = find_project_root()
shared_notebook = PROJECT_ROOT / "notebooks" / "_Shared_Training_Utilities.ipynb"
get_ipython().run_line_magic("run", str(shared_notebook))
```

```python
config = EXPERIMENT_CONFIG["experiment_5_multibranch_multitask"]
metrics, predictions, history, config_out = run_multibranch_experiment(config, seed=42, save_model=True)
```
