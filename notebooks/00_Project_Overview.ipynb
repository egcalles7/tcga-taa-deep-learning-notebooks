# TCGA TAA Deep Learning Project Overview

This notebook is a narrative overview. The actual pipeline is organized across notebooks 01–07.

## Research question

Can neural network architectures predict tumor-associated antigen RNA expression from TCGA multi-omic features?

The response variables are 10 TAA RNA expression values. The features include cancer type/clinical phenotype, selected driver-gene mutation features, and gene-level copy-number features.

## Deep learning emphasis

This project focuses on:

- feedforward neural networks
- backpropagation
- regularization with dropout
- batch normalization
- residual connections
- modality-specific representation learning
- multi-task prediction with gene-specific output heads

## Experiment plan

1. Standard MLP
2. MLP without dropout vs with dropout
3. MLP with batch normalization vs without batch normalization
4. Residual MLP
5. Multi-branch multi-task neural network
